```python
sayi1 = 12
sayi2 = 5

print("Toplam:", sayi1 + sayi2)
print("Fark:", sayi1 - sayi2)
print("Çarpım:", sayi1 * sayi2)
print("Bölüm:", sayi1 / sayi2)
print("Kalan:", sayi1 % sayi2)
print("Üs alma:", sayi1 ** sayi2)
print("Tam bölme:", sayi1 // sayi2)
```

```python
# Atama Operatörleri
sayi = 50
print(sayi)

sayi += 5 # sayı = sayı + 5
print(sayi)

sayi -= 3 # sayı = sayı - 3
print(sayi)

sayi *= 2 # sayı = sayı * 2
print(sayi)

sayi /= 4 # sayı = sayı / 4
print(sayi)

sayi %= 3 # sayı = sayı % 3
print(sayi)

sayi **= 2 # sayı = sayı ** 2
print(sayi)

sayi //= 2 # sayı = sayı // 2
print(sayi)
```
