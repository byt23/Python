```python
def durum_kontrol(kod):
    match kod:
        case 200:
            return "Başarılı : Bağlantı kuruldu."
        case 400 | 401 | 403 | 404 : # Pipe (|) : birden fazla değeri aynı blokta yakalayabiliyoruz. 
            return "Hata : İstemci kaynanklı bir problem var."
        case 500 | 502 | 503 :
            return "Hata : Sunucu kaynaklı bir problem var."
        case _ :
            return "Bilinmeyen durum kodu."

print(durum_kontrol(357))
print(durum_kontrol(401))
```

```python
def sinav_sonucu_degerlendir(ogrenci_verisi):
    match ogrenci_verisi:
        case (uid, puan) if puan >= 90:
            return f"{uid} numaralı öğrenci: Üstün Başarı! Puanı: {puan}"
        case (uid, puan) if 50 <= puan < 90:
            return f"{uid} numaralı öğrenci: Başarılı. Puanı: {puan}"
        case (uid, puan) if puan < 50:
            return f"{uid} numaralı öğrenci: Başarısız. Ek çalışma programı oluşturulmalı."
        case _:
            return "Geçersiz veri formatı!"

print(sinav_sonucu_degerlendir(("YKS-2026-001", 95))) 
```

```python
def vites(vites_durumu):
    match vites_durumu:
        case "P":
            return "Park"
        case "R":
            return "Geri"
        case "N":
            return "Boş"
        case "D":
            return "İleri"
        case _:
            return "Şanzımanı ele aldın"

print(vites(("P")))
print(vites(("R")))
print(vites(("N")))
print(vites(("D")))
print(vites(("A")))
```

```python
def koordinat_bul(koordinat):
    match koordinat:
        case (0,0) :
            return "Orijin"
        case (0,5) :
            return "Y ekseni üzerinde"
        case (7,0) :
            return "X ekseni üzerinde"
        case _:
            return f"Düzlem ({x,y}) üzerinde"

print(koordinat_bul((0,5)))
```
